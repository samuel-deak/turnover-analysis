```python
# Importação de bibliotecas necessárias
import pandas as pd
import nbformat as nbf
import ipykernel
import plotly.express as px
import openpyxl
```

```python
#Importar e visualizar a base de dados
tabela = pd.read_csv("Base de dados.csv")
display(tabela)
```

```python
# Análise de dados faltantes
print(tabela.info())
```

```python
# Remover colunas desnecessárias
tabela = tabela.drop(columns=["EmployeeNumber", "EmployeeCount", "Over18", "StandardHours"])
display(tabela)
```

```python
# Análise e visualização dos dados iniciais
lista = tabela["Attrition"].value_counts(normalize=True)
print(lista)
```

```python
#Análise detalhada (Padrões de demissão da empresa)
for coluna in tabela.columns:
    grafico = px.histogram(tabela, x=coluna, color="Attrition", barmode="group", barnorm="percent")
    grafico.show()
```

```python
#Gasto anual com os funcionários
gasto_total = tabela["MonthlyIncome"].sum() * 12
print("O gasto anual com os funcionários é: R$", gasto_total) 
#Gasto anual de reposição de funcionários devido à demissões
gasto_anual = tabela[tabela["Attrition"] == "Yes"]["MonthlyIncome"].sum() * 12 * 1.5
print("O gasto anual de reposição de funcionários é: R$", gasto_anual)
```
